```python
import pandas as pd
import numpy as np

data = pd.read_csv("diabetes.csv")

print("Column Names:")
print(list(data.columns))

print("\nData Types:")
print(data.dtypes)

print("\nTotal Records:", data.shape[0])

print("\nFirst 20 Records:")
print(data.iloc[:20])

print("\nLast 20 Records:")
print(data.iloc[-20:])

data = data.rename(columns={"Outcome": "Diagnosis"})

data["Classification"] = np.where(
    data["Diagnosis"] == 1,
    "Diabetes",
    "No Diabetes"
)

withDiabetes = data.query("Diagnosis == 1")
noDiabetes = data.query("Diagnosis == 0")

Pedia = data[data["Age"] <= 19]
Adult = data[data["Age"] > 19]

avg_age = data["Age"].mean()
avg_glucose = data["Glucose"].mean()

print("\nAverage Age:", avg_age)
print("Average Glucose:", avg_glucose)

median_age = data["Age"].median()
median_glucose = data["Glucose"].median()

print("\nMedian Age:", median_age)
print("Median Glucose:", median_glucose)

middle_age = np.percentile(data["Age"], 50)
middle_glucose = np.percentile(data["Glucose"], 50)

print("\nMiddle Age:", middle_age)
print("Middle Glucose:", middle_glucose)

std_skin = data["SkinThickness"].std()

print("\nStandard Deviation of SkinThickness:", std_skin)
```
