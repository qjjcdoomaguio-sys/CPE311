**Example of gathering  image data using webcam** <br>
Note: Run this snippet using local jupyter notebook

```python
import cv2

webcam = cv2.VideoCapture(0)

print("Press 's' to capture image")
print("Press 'q' to quit")

while True:
    try:

        check, frame = webcam.read()
        print(check)

        if not check:
            print("Failed to access webcam")
            break


        cv2.imshow("Capturing", frame)

        key = cv2.waitKey(1)


        if key == ord('s'):
            cv2.imwrite('saved_img.jpg', frame)
            print("Image saved!")

            print("Converting RGB image to grayscale...")
            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

            print("Resizing image to 28x28...")
            resized = cv2.resize(gray, (28, 28))

            cv2.imwrite('saved_img-final.jpg', resized)
            print("Processed image saved!")

            cv2.imshow("Captured Image", resized)
            cv2.waitKey(1650)

            break

        elif key == ord('q'):
            print("Turning off camera.")
            break

    except KeyboardInterrupt:
        print("Interrupted by user.")
        break

webcam.release()
cv2.destroyAllWindows()
print("Camera off. Program ended.")
```

**Image Scraping using Selenium** <br>
Note: Run the snippet of code using local jupyter notebook

```python
!pip install selenium webdriver-manager requests

import time
import requests
import os
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.service import Service

# Chrome options
chrome_options = Options()
chrome_options.add_argument("--headless=new")

# Start driver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=chrome_options)

def get_image_urls(query, total_imgs):
    search_url = f"https://www.google.com/search?q={query}&tbm=isch"
    driver.get(search_url)
    time.sleep(2)

    img_urls = set()

    while len(img_urls) < total_imgs:
        thumbnails = driver.find_elements(By.CSS_SELECTOR, "img.Q4LuWd")

        for img in thumbnails:
            try:
                img.click()
                time.sleep(1)

                images = driver.find_elements(By.CSS_SELECTOR, "img.n3VNCb")
                for image in images:
                    src = image.get_attribute("src")
                    if src and src.startswith("http"):
                        img_urls.add(src)

                if len(img_urls) >= total_imgs:
                    break
            except:
                continue

        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)

        print(f"Collected: {len(img_urls)} images")

    return list(img_urls)[:total_imgs]


def download_images(folder, urls):
    if not os.path.exists(folder):
        os.makedirs(folder)

    for i, url in enumerate(urls):
        try:
            img_data = requests.get(url, timeout=10).content
            with open(os.path.join(folder, f"image_{i}.jpg"), "wb") as f:
                f.write(img_data)
        except Exception as e:
            print(f"Failed: {e}")

# Run
urls = get_image_urls("Car", 10)
download_images("car_images", urls)

driver.quit()
print("Done!")
```

**Web Scraping of Movies Information using BeautifulSoup** <br>
We want to analyze the distributions of IMDB and Metacritic movie ratings to see if we find anything interesting. To do this, weʼll first scrape data for over 2000 movies.

note: Top 50 Movies in 2000


```python
from requests import get
url = 'https://www.imdb.com/search/title?release_date=2017&sort=num_votes,desc&page=1'
response = get(url)
print(response.text[:500])
```

```python
from bs4 import BeautifulSoup
html_soup = BeautifulSoup(response.text, 'html.parser')
headers = {'Accept-Language': 'en-US,en;q=0.8'}
type(html_soup)
```

```python
movie_containers = html_soup.find_all('div', class_ = 'lister-item mode-advanced')
print(type(movie_containers))
print(len(movie_containers))
```
