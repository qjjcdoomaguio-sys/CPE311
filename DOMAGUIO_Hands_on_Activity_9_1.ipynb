Creating subplots

```python
fb.plot(
kind='line',
subplots=True,
layout=(3,2),
figsize=(15,10),
title='Facebook Stock 2018'
)
```

```python
fb.assign(
max_abs_change=fb.high - fb.low
).plot(
kind='scatter', x='volume', y='max_abs_change',
title='Facebook Daily High - Low vs. Volume Traded'
)
```

```python
fb.assign(
max_abs_change=fb.high - fb.low
).plot(
kind='scatter', x='volume', y='max_abs_change',
title='Facebook Daily High - Low vs. log(Volume Traded)',
logx=True
)
```

```python
fb.assign(
max_abs_change=fb.high - fb.low
).plot(
kind='scatter', x='volume', y='max_abs_change',
title='Facebook Daily High - Low vs. log(Volume Traded)',
logx=True, alpha=0.25
)
```

```python
fb.assign(
log_volume=np.log(fb.volume),
max_abs_change=fb.high - fb.low
).plot(
kind='hexbin',
x='log_volume',
y='max_abs_change',
title='Facebook Daily High - Low vs. log(Volume Traded)',
colormap='gray_r',
gridsize=20,
sharex=False
)
```

```python
fig, ax = plt.subplots(figsize=(20, 10))
fb_corr = fb.assign(
    log_volume=np.log(fb.volume),
    max_abs_change=fb.high - fb.low
).corr()

im = ax.matshow(fb_corr, cmap='seismic')

cbar = fig.colorbar(im)
im.set_clim(-1, 1)

labels = [col.lower() for col in fb_corr.columns]
ax.set_xticks(range(len(labels)))
ax.set_xticklabels(labels, rotation=45)
ax.set_yticks(range(len(labels)))
ax.set_yticklabels(labels)
```

Stacked bar

```python
pivot = quakes.assign(
mag_bin=lambda x: np.floor(x.mag)
).pivot_table(
index='mag_bin', columns='magType', values='mag', aggfunc='count'
)
pivot.plot.bar(
stacked=True, rot=0,
title='Earthquakes by integer magnitude and magType'
)
plt.ylabel('earthquakes') # label the axes (discussed in chapter 6)
```

Normalized stacked bars

```python
normalized_pivot = pivot.fillna(0).apply(lambda x: x/x.sum(), axis=1)
ax = normalized_pivot.plot.bar(
stacked=True, rot=0, figsize=(10, 5),
title='Percentage of earthquakes by integer magnitude for each magType'
)
ax.legend(bbox_to_anchor=(1, 0.8)) # move legend to the right of the plot
plt.ylabel('percentage')
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

fb = pd.read_csv('fb_stock_prices_2018.csv', index_col='date', parse_dates=True)
quakes = pd.read_csv('earthquakes-1.csv')
```

1. Plot the rolling 20-day minimum of the Facebook closing price with the pandas plot() method

```python
fb.close.rolling(20).min().plot(
    figsize=(10,5),
    title='Rolling 20-Day Minimum of Facebook Closing Price'
)
plt.xlabel('date')
plt.ylabel('price')
plt.show()
```

2. Create a histogram and KDE of the change from open to close in the price of Facebook stock.

```python
change = fb.close - fb.open
ax = change.plot(
    kind='hist',
    density=True,
    alpha=0.5,
    figsize=(10,5),
    title='Histogram and KDE of Change from Open to Close'
)
change.plot(kind='kde', ax=ax)
plt.xlabel('price change')
plt.show()
```

3. Using the earthquake data, create box plots for the magnitudes of each magType used in Indonesia.

```python
indo = quakes.query('parsed_place == "Indonesia"')
indo[['mag','magType']].groupby('magType').boxplot(
    figsize=(12,6),
    subplots=False
)
plt.title('Earthquake Magnitudes by magType in Indonesia')
plt.ylabel('magnitude')
plt.show()
```

4. Make a line plot of the difference between the weekly maximum high price and the weekly minimum low price for Facebook. This should be a single line.



```python
weekly_diff = fb.high.resample('W').max() - fb.low.resample('W').min()
weekly_diff.plot(
    figsize=(10,5),
    title='Weekly Max High - Weekly Min Low (Facebook)'
)
plt.xlabel('date')
plt.ylabel('price difference')
plt.show()
```

5. Using matplotlib and pandas, create two subplots side-by-side showing the effect that after-hours trading has had on Facebook's stock price: <br>
  
  Bonus #1: Color the bars according to whether they are gains in the stock price (green) or drops in the stock price (red).

  Bonus #2: Modify the x-axis of the bar plot to show the threeletter abbreviation for the month.

```python
daily_diff = fb.open - fb.close.shift(1)

fig, axes = plt.subplots(1, 2, figsize=(15,5))

daily_diff.plot(
    ax=axes[0],
    title='Daily Difference: Open - Prior Close'
)
axes[0].set_xlabel('date')
axes[0].set_ylabel('price difference')

monthly_effect = daily_diff.resample('M').sum()
colors = ['green' if x > 0 else 'red' for x in monthly_effect]

monthly_effect.index = monthly_effect.index.strftime('%b')

monthly_effect.plot(
    kind='bar',
    ax=axes[1],
    color=colors,
    title='Monthly Net Effect of After-Hours Trading'
)
axes[1].set_xlabel('month')
axes[1].set_ylabel('net price difference')
axes[1].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```
