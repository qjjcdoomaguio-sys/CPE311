```python
# Initial setup
n = 4
rods = {
    'A': [4, 3, 2, 1],
    'B': [],
    'C': []
}

def print_rods():
    print(f"A = {rods['A']}")
    print(f"B = {rods['B']}")
    print(f"C = {rods['C']}")
    print()

def TowerOfHanoi(n, src, aux, des):
    if n == 1:
        disk = rods[src].pop()
        rods[des].append(disk)
        print(f"Move disk {disk} from rod {src} → rod {des}")
        print_rods()
        return

    TowerOfHanoi(n - 1, src, des, aux)
    disk = rods[src].pop()
    rods[des].append(disk)
    print(f"Move disk {disk} from rod {src} → rod {des}")
    print_rods()

    TowerOfHanoi(n - 1, aux, src, des)

#stage 1
print("Stage 1")
print("source: A")
print("Aux: B")
print("Dest: C")
print("Stopping condition: All in B Except disk 4\n")

# Move 3 disks from A to B using C
TowerOfHanoi(3, 'A', 'C', 'B')

# stage 2
print("Stage 2")
print("source: B")
print("Aux: A")
print("Dest: C")
print("Stopping condition: All in rod C\n")

# Move disk 4
disk = rods['A'].pop()
rods['C'].append(disk)
print(f"Move disk {disk} from rod A → rod C")
print_rods()

# Move 3 disks from B to C using A
TowerOfHanoi(3, 'B', 'A', 'C')
```
